Problem 1. Create a definition that will randomly pick vectors, given a user defined number of clusters, K, and the sample data.  Return K random points contained within data

```python
#random starting points
def initialize_centroids(data, K):

  #random shuffle indices
  indices = np.arange(len(data))
  np.random.shuffle(indices)

  selected_indices = indices[:K]

  #get respective vectors
  centroids = data[selected_indices]

  return centroids


K = 3  # Number of clusters
centroids = initialize_centroids(data, K)
print(centroids)
```

Problem 2. Create a defintion that will calculate the euclidean distance between each centroid and each data point and assign each data point to the nearest cluster.  Return a list, where the outer list index is equivical to number of clusters K, inner list is assignment of each vector from data to its respective cluster

```python
#assign each vector to a cluster based on Euclidean distances
def assign_to_clusters(data, centroids):

  K = centroids.shape[0]

  num_data_points = data.shape[0]

  #initialize list to store assignments
  assignments = [[] for _ in range(K)]

  #iterate through data points
  for i in range(num_data_points):

    #calculate the Euclidean distance between data point and each centroid
    distances = np.linalg.norm(centroids - data[i], axis=1)

    #find the index of the centroid with the minimum distance
    nearest_cluster_index = np.argmin(distances)

    #assign data point to the nearest cluster
    assignments[nearest_cluster_index].append(i)

  #print out the list of assignments
  print("Cluster Assignments:")
  for cluster_index, data_indices in enumerate(assignments):
    print(f"Cluster {cluster_index + 1}: {data_indices}")

  return assignments

assignments = assign_to_clusters(data, centroids)
```

Problem 3. Create a definition to find the average of each constituent vector assigned to a cluster and return mean for each cluster.  return cluster means for each cluster K

```python
#compute each cluster as average of its constituents
def calculate_cluster_means(data, assignments):

  K = len(assignments)

  #initialize array to store cluster means
  cluster_means = np.zeros((K, data.shape[1]))


  for i in range(K):
    #get indices of data points assigned to this cluster
    cluster_indices = assignments[i]

    #extract data points assigned to this cluster
    cluster_data = data[cluster_indices]

    #calculate cluster mean
    cluster_mean = np.mean(cluster_data, axis=0)

    cluster_means[i] = cluster_mean

  return cluster_means
```

Problem 4. Create a definition to cluster each data point until means do not change, i.e. the difference is equal to 0. Return the clusters and nested list with assignment of each vector to each cluster K

```python
#iterate clusering
def k_means_clustering(data, K):
  #randomly initialize K centroids
  centroids = data[np.random.choice(data.shape[0], K, replace=False)]

  #initialize old centroids to compare changes
  old_centroids = np.zeros_like(centroids)

  #assign each data point to the nearest cluster centroid
  while not np.array_equal(centroids, old_centroids):
    old_centroids = centroids.copy()

    #calculate the Euclidean distance between data points and each centroid
    distances = np.linalg.norm(data[:, np.newaxis] - centroids, axis=2)

    #assign each data point to nearest centroid
    assignments = np.argmin(distances, axis=1)

    #update centroids based on the mean of data points in each cluster
    for i in range(K):
      centroids[i] = np.mean(data[assignments == i], axis=0)

  return centroids, [np.where(assignments == i)[0] for i in range(K)]

centroids, clusters = k_means_clustering(data, K)
print("Clusters:")
for i, cluster in enumerate(clusters):
    print(f"Cluster {i + 1}: {cluster}")
print("Centroids:")
print(centroids)
```
